````python
import boto3
import json
import os
import time
import logging
import hashlib
import pickle
import gradio as gr
from pathlib import Path
from typing import Dict, List, Any, Tuple, Optional

# Configure logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
logger = logging.getLogger("bedrock-summarizer")

# Set AWS region and initialize Bedrock client
AWS_REGION = 'us-east-1'  # Make sure to use a region where Bedrock is available

try:
    session = boto3.Session(region_name=AWS_REGION)
    credentials = session.get_credentials()
    
    if credentials is None:
        print("⚠️ AWS credentials not found. Please configure your credentials.")
    else:
        print(f"✓ Using AWS profile: {session.profile_name or 'default'}")
        print(f"✓ Using region: {session.region_name}")
    
    # First, make sure boto3 is up-to-date
    import pkg_resources
    boto3_version = pkg_resources.get_distribution("boto3").version
    print(f"✓ Using boto3 version: {boto3_version}")
    
    # Try different approaches to initialize Bedrock client
    try:
        # Approach 1: Direct service name (requires boto3 >= 1.28.0)
        bedrock = boto3.client(
            service_name='bedrock-runtime',
            region_name=AWS_REGION
        )
        print("✓ Bedrock client initialized successfully using 'bedrock-runtime' service name")
    except Exception as e1:
        # Approach 2: Using endpoint URL
        try:
            bedrock = boto3.client(
                service_name='runtime.bedrock',  # Alternative name
                region_name=AWS_REGION
            )
            print("✓ Bedrock client initialized successfully using 'runtime.bedrock' service name")
        except Exception as e2:
            # Approach 3: Using a custom endpoint URL
            try:
                bedrock = boto3.client(
                    service_name='bedrock',
                    region_name=AWS_REGION,
                    endpoint_url=f'https://bedrock-runtime.{AWS_REGION}.amazonaws.com'
                )
                print("✓ Bedrock client initialized successfully using custom endpoint URL")
            except Exception as e3:
                print(f"❌ Error with approach 1: {str(e1)}")
                print(f"❌ Error with approach 2: {str(e2)}")
                print(f"❌ Error with approach 3: {str(e3)}")
                print("\n⚠️ IMPORTANT: You may need to update boto3 with:")
                print("!pip install --upgrade boto3")
                print("\nOr install boto3 extras for Bedrock:")
                print("!pip install 'boto3>=1.28.0'")
                raise Exception("Could not initialize Bedrock client")

    # Verify that the Bedrock client works by listing models
    try:
        # Create a regular bedrock client (not runtime) to check model access
        bedrock_mgmt = boto3.client(
            service_name='bedrock',
            region_name=AWS_REGION
        )
        models = bedrock_mgmt.list_foundation_models()
        print(f"✓ Successfully connected to Bedrock - found {len(models.get('modelSummaries', []))} models")
    except Exception as e:
        print(f"⚠️ Could not list models: {str(e)}")
        print("  This may be because you need to enable Bedrock in your AWS account")
        print("  Visit: https://console.aws.amazon.com/bedrock/ to enable it")
        # Continuing anyway as the issue might be with permissions
        
except Exception as e:
    print(f"❌ Error initializing Bedrock client: {str(e)}")

# Define constants
CLAUDE_MODEL_ID = "anthropic.claude-3-sonnet-20240229-v1:0"
TITAN_EMBEDDINGS_MODEL_ID = "amazon.titan-embed-text-v1"
CACHE_DIR = "embeddings_cache"
CACHE_FILE = os.path.join(CACHE_DIR, "embeddings_cache.pkl")

# Create cache directory
Path(CACHE_DIR).mkdir(parents=True, exist_ok=True)
print(f"✓ Using Claude model: {CLAUDE_MODEL_ID}")
print(f"✓ Using Embeddings model: {TITAN_EMBEDDINGS_MODEL_ID}")
print(f"✓ Cache directory: {CACHE_DIR}")
print(f"✓ Cache file: {CACHE_FILE}")

# Initialize global cache
embedding_cache = {}
````

````python


## Cell 2: Embedding Cache Functions

def get_text_hash(text: str) -> str:
    """Generate a deterministic hash for the text"""
    return hashlib.md5(text.encode('utf-8')).hexdigest()

def load_embedding_cache() -> Dict[str, List[float]]:
    """Load the embedding cache from file"""
    global embedding_cache
    
    if embedding_cache:
        logger.info("Using already loaded embedding cache")
        return embedding_cache
        
    if os.path.exists(CACHE_FILE):
        logger.info(f"Loading embedding cache from {CACHE_FILE}")
        try:
            with open(CACHE_FILE, 'rb') as f:
                embedding_cache = pickle.load(f)
            logger.info(f"Loaded cache with {len(embedding_cache)} entries")
            return embedding_cache
        except Exception as e:
            logger.warning(f"Failed to load embedding cache: {str(e)}")
            embedding_cache = {}
    else:
        logger.info(f"No cache file found at {CACHE_FILE}, starting with empty cache")
        embedding_cache = {}
    
    return embedding_cache

def save_embedding_cache() -> bool:
    """Save the embedding cache to file"""
    try:
        with open(CACHE_FILE, 'wb') as f:
            pickle.dump(embedding_cache, f)
        logger.info(f"Saved embedding cache with {len(embedding_cache)} entries to {CACHE_FILE}")
        return True
    except Exception as e:
        logger.error(f"Failed to save embedding cache: {str(e)}")
        return False

def clear_cache():
    """Clear the embedding cache"""
    global embedding_cache
    try:
        if os.path.exists(CACHE_FILE):
            os.remove(CACHE_FILE)
            print(f"✓ Cache file deleted: {CACHE_FILE}")
        else:
            print("ℹ No cache file to delete")
            
        embedding_cache = {}
        print("✓ In-memory cache cleared")
        
    except Exception as e:
        print(f"❌ Error clearing cache: {str(e)}")

def show_cache_stats():
    """Show statistics about the embedding cache"""
    cache = load_embedding_cache()
    print("\n📊 CACHE STATISTICS")
    print("-----------------")
    print(f"Total entries: {len(cache)}")
    
    if os.path.exists(CACHE_FILE):
        file_size = os.path.getsize(CACHE_FILE) / 1024  # KB
        print(f"File size: {file_size:.2f} KB")
    
    if cache:
        sample_key = next(iter(cache))
        sample_embedding = cache[sample_key]
        print(f"Embedding dimension: {len(sample_embedding)}")
    else:
        print("Cache is empty")

def get_cache_stats_text():
    """Return cache statistics as a formatted string"""
    cache = load_embedding_cache()
    stats = f"Total entries: {len(cache)}\n"
    
    if os.path.exists(CACHE_FILE):
        file_size = os.path.getsize(CACHE_FILE) / 1024  # KB
        stats += f"File size: {file_size:.2f} KB\n"
    
    if cache:
        sample_key = next(iter(cache))
        sample_embedding = cache[sample_key]
        stats += f"Embedding dimension: {len(sample_embedding)}"
    else:
        stats += "Cache is empty"
    
    return stats

````

````python


## Cell 3: Core Embedding and Summarization Functions

def get_embeddings(text: str) -> Tuple[List[float], bool]:
    """Generate or retrieve embeddings for the given text"""
    text_hash = get_text_hash(text)
    
    # Load cache if needed
    cache = load_embedding_cache()
    
    # Check if text hash exists in cache
    if text_hash in cache:
        logger.info(f"Found cached embeddings for text hash: {text_hash}")
        return cache[text_hash], True
    
    # If not in cache, generate new embeddings
    start_time = time.time()
    logger.info("Starting new embedding generation with Titan Embeddings model")
    
    try:
        logger.info(f"Preparing request for text of length: {len(text)} characters")
        response = bedrock.invoke_model(
            modelId=TITAN_EMBEDDINGS_MODEL_ID,
            body=json.dumps({
                "inputText": text
            })
        )
        logger.info("Received response from Titan Embeddings model")
        
        response_body = json.loads(response['body'].read().decode('utf-8'))
        embeddings = response_body['embedding']
        
        end_time = time.time()
        logger.info(f"Embedding generation completed in {end_time - start_time:.2f} seconds")
        logger.info(f"Generated embeddings with dimension: {len(embeddings)}")
        
        # Store in cache
        cache[text_hash] = embeddings
        if save_embedding_cache():
            logger.info("Embeddings successfully cached for future use")
        
        return embeddings, False
    except Exception as e:
        logger.error(f"Error generating embeddings: {str(e)}")
        raise

def generate_summary(text: str, query: str) -> str:
    """Generate summary using Claude model"""
    start_time = time.time()
    logger.info(f"Starting summary generation with Claude model")
    logger.info(f"Query: '{query}'")
    logger.info(f"Input text length: {len(text)} characters")
    
    try:
        prompt = f"""
        I need you to summarize the following text based on this query: "{query}"
        
        Text to summarize:
        {text}
        
        Please provide a concise and relevant summary focusing on aspects related to the query.
        """
        
        logger.info("Preparing request for Claude model")
        response = bedrock.invoke_model(
            modelId=CLAUDE_MODEL_ID,
            body=json.dumps({
                "anthropic_version": "bedrock-2023-05-31",
                "max_tokens": 1000,
                "temperature": 0,
                "messages": [
                    {
                        "role": "user",
                        "content": prompt
                    }
                ]
            })
        )
        logger.info("Received response from Claude model")
        
        response_body = json.loads(response['body'].read().decode('utf-8'))
        summary = response_body['content'][0]['text']
        
        end_time = time.time()
        logger.info(f"Summary generation completed in {end_time - start_time:.2f} seconds")
        logger.info(f"Generated summary length: {len(summary)} characters")
        
        return summary
    except Exception as e:
        logger.error(f"Error generating summary: {str(e)}")
        raise

def process_text(text: str, query: str) -> Dict[str, Any]:
    """Process text to generate embeddings and summary"""
    overall_start_time = time.time()
    logger.info("Processing started")
    
    result = {}
    
    try:
        logger.info(f"Processing request with query: '{query}'")
        logger.info(f"Input text length: {len(text)} characters")
        
        if not text or not query:
            logger.error("Missing text or query parameter")
            return {'error': 'Missing text or query parameter'}
        
        # Generate or retrieve embeddings for the text
        logger.info("Starting embedding process")
        embeddings, from_cache = get_embeddings(text)
        logger.info(f"Embeddings {'retrieved from cache' if from_cache else 'generated'} with dimension: {len(embeddings)}")
        
        # Generate summary based on the query
        logger.info("Starting summary generation process")
        summary = generate_summary(text, query)
        logger.info("Summary successfully generated")
        
        overall_end_time = time.time()
        processing_time = overall_end_time - overall_start_time
        logger.info(f"Total processing time: {processing_time:.2f} seconds")
        
        result = {
            'summary': summary,
            'embedding_dimension': len(embeddings),
            'embedding_source': 'cache' if from_cache else 'generated',
            'processing_time_seconds': round(processing_time, 2)
        }
        
    except Exception as e:
        logger.error(f"Error in processing: {str(e)}")
        result = {'error': str(e)}
    
    return result
````

````python

def get_text_hash(text: str) -> str:
    """Generate a deterministic hash for the text"""
    return hashlib.md5(text.encode('utf-8')).hexdigest()

def load_embedding_cache() -> Dict[str, List[float]]:
    """Load the embedding cache from file"""
    global embedding_cache
    
    if embedding_cache:
        logger.info("Using already loaded embedding cache")
        return embedding_cache
        
    if os.path.exists(CACHE_FILE):
        logger.info(f"Loading embedding cache from {CACHE_FILE}")
        try:
            with open(CACHE_FILE, 'rb') as f:
                embedding_cache = pickle.load(f)
            logger.info(f"Loaded cache with {len(embedding_cache)} entries")
            return embedding_cache
        except Exception as e:
            logger.warning(f"Failed to load embedding cache: {str(e)}")
            embedding_cache = {}
    else:
        logger.info(f"No cache file found at {CACHE_FILE}, starting with empty cache")
        embedding_cache = {}
    
    return embedding_cache

def save_embedding_cache() -> bool:
    """Save the embedding cache to file"""
    try:
        with open(CACHE_FILE, 'wb') as f:
            pickle.dump(embedding_cache, f)
        logger.info(f"Saved embedding cache with {len(embedding_cache)} entries to {CACHE_FILE}")
        return True
    except Exception as e:
        logger.error(f"Failed to save embedding cache: {str(e)}")
        return False

def clear_cache():
    """Clear the embedding cache"""
    global embedding_cache
    try:
        if os.path.exists(CACHE_FILE):
            os.remove(CACHE_FILE)
            print(f"✓ Cache file deleted: {CACHE_FILE}")
        else:
            print("ℹ No cache file to delete")
            
        embedding_cache = {}
        print("✓ In-memory cache cleared")
        
    except Exception as e:
        print(f"❌ Error clearing cache: {str(e)}")

def show_cache_stats():
    """Show statistics about the embedding cache"""
    cache = load_embedding_cache()
    print("\n📊 CACHE STATISTICS")
    print("-----------------")
    print(f"Total entries: {len(cache)}")
    
    if os.path.exists(CACHE_FILE):
        file_size = os.path.getsize(CACHE_FILE) / 1024  # KB
        print(f"File size: {file_size:.2f} KB")
    
    if cache:
        sample_key = next(iter(cache))
        sample_embedding = cache[sample_key]
        print(f"Embedding dimension: {len(sample_embedding)}")
    else:
        print("Cache is empty")
````

````python

## Cell 3: Core Embedding and Summarization Functions

def get_embeddings(text: str) -> Tuple[List[float], bool]:
    """Generate or retrieve embeddings for the given text"""
    text_hash = get_text_hash(text)
    
    # Load cache if needed
    cache = load_embedding_cache()
    
    # Check if text hash exists in cache
    if text_hash in cache:
        logger.info(f"Found cached embeddings for text hash: {text_hash}")
        return cache[text_hash], True
    
    # If not in cache, generate new embeddings
    start_time = time.time()
    logger.info("Starting new embedding generation with Titan Embeddings model")
    
    try:
        logger.info(f"Preparing request for text of length: {len(text)} characters")
        response = bedrock.invoke_model(
            modelId=TITAN_EMBEDDINGS_MODEL_ID,
            body=json.dumps({
                "inputText": text
            })
        )
        logger.info("Received response from Titan Embeddings model")
        
        response_body = json.loads(response['body'].read().decode('utf-8'))
        embeddings = response_body['embedding']
        
        end_time = time.time()
        logger.info(f"Embedding generation completed in {end_time - start_time:.2f} seconds")
        logger.info(f"Generated embeddings with dimension: {len(embeddings)}")
        
        # Store in cache
        cache[text_hash] = embeddings
        if save_embedding_cache():
            logger.info("Embeddings successfully cached for future use")
        
        return embeddings, False
    except Exception as e:
        logger.error(f"Error generating embeddings: {str(e)}")
        raise

def generate_summary(text: str, query: str) -> str:
    """Generate summary using Claude model"""
    start_time = time.time()
    logger.info(f"Starting summary generation with Claude model")
    logger.info(f"Query: '{query}'")
    logger.info(f"Input text length: {len(text)} characters")
    
    try:
        prompt = f"""
        I need you to summarize the following text based on this query: "{query}"
        
        Text to summarize:
        {text}
        
        Please provide a concise and relevant summary focusing on aspects related to the query.
        """
        
        logger.info("Preparing request for Claude model")
        response = bedrock.invoke_model(
            modelId=CLAUDE_MODEL_ID,
            body=json.dumps({
                "anthropic_version": "bedrock-2023-05-31",
                "max_tokens": 1000,
                "temperature": 0,
                "messages": [
                    {
                        "role": "user",
                        "content": prompt
                    }
                ]
            })
        )
        logger.info("Received response from Claude model")
        
        response_body = json.loads(response['body'].read().decode('utf-8'))
        summary = response_body['content'][0]['text']
        
        end_time = time.time()
        logger.info(f"Summary generation completed in {end_time - start_time:.2f} seconds")
        logger.info(f"Generated summary length: {len(summary)} characters")
        
        return summary
    except Exception as e:
        logger.error(f"Error generating summary: {str(e)}")
        raise

def process_text(text: str, query: str) -> Dict[str, Any]:
    """Process text to generate embeddings and summary"""
    overall_start_time = time.time()
    logger.info("Processing started")
    
    result = {}
    
    try:
        logger.info(f"Processing request with query: '{query}'")
        logger.info(f"Input text length: {len(text)} characters")
        
        if not text or not query:
            logger.error("Missing text or query parameter")
            return {'error': 'Missing text or query parameter'}
        
        # Generate or retrieve embeddings for the text
        logger.info("Starting embedding process")
        embeddings, from_cache = get_embeddings(text)
        logger.info(f"Embeddings {'retrieved from cache' if from_cache else 'generated'} with dimension: {len(embeddings)}")
        
        # Generate summary based on the query
        logger.info("Starting summary generation process")
        summary = generate_summary(text, query)
        logger.info("Summary successfully generated")
        
        overall_end_time = time.time()
        processing_time = overall_end_time - overall_start_time
        logger.info(f"Total processing time: {processing_time:.2f} seconds")
        
        result = {
            'summary': summary,
            'embedding_dimension': len(embeddings),
            'embedding_source': 'cache' if from_cache else 'generated',
            'processing_time_seconds': round(processing_time, 2)
        }
        
    except Exception as e:
        logger.error(f"Error in processing: {str(e)}")
        result = {'error': str(e)}
    
    return result
````
